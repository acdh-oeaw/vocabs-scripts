```python
from rdflib import Graph
from pprint import pprint
import pandas as pd

# Load the Turtle RDF file into a graph
g = Graph()
g.parse("arche_licenses.ttl", format="turtle")

print(f"Loaded {len(g)} triples.")
```

```python
query = """
SELECT ?s ?prefLabel ?altLabel
WHERE {
    ?s rdf:type skos:Concept .
    OPTIONAL { ?s skos:prefLabel ?prefLabel . FILTER (lang(?prefLabel) = "en") }
    OPTIONAL { ?s skos:altLabel ?altLabel . FILTER (lang(?altLabel) = "en") }
}
"""

concepts = []

qres = g.query(query)
for row in qres:
    concept = {"uri": str(row[0]),
               "prefLabel": str(row[1]),
               "altLabel": str(row[2])}
    concepts.append(concept)

pprint(concepts)
```

```python
df = pd.DataFrame.from_dict(concepts)
df.to_csv("arche_licenses_concepts.csv", index=False)
```
